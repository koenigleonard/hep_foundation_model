```python
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.ticker import MaxNLocator
import numpy as np

jetclass_qcd_linear = pd.read_csv("/hpcwork/rwth0934/hep_foundation_model/checkpoints/JETCLASS_QCD_600000_LINEAR_training_log.csv")
jetclass_top_linear = pd.read_csv("/hpcwork/rwth0934/hep_foundation_model/checkpoints/JETCLASS_TTBar_600000_LINEAR_training_log.csv")

jetclass_qcd_factorized = pd.read_csv("/hpcwork/rwth0934/hep_foundation_model/checkpoints/JETCLASS_QCD_600000_FACTORIZED_training_log.csv")
jetclass_top_factorized = pd.read_csv("/hpcwork/rwth0934/hep_foundation_model/checkpoints/JETCLASS_TTBar_600000_FACTORIZED_training_log.csv")
```

```python
top_linear_min = np.argmin(jetclass_top_linear["val_loss"])
top_factorized_min = np.argmin(jetclass_top_factorized["val_loss"])

qcd_linear_min = np.argmin(jetclass_qcd_linear["val_loss"])
qcd_factorized_min = np.argmin(jetclass_qcd_factorized["val_loss"])
```

```python

fig, ax = plt.subplots(figsize=(10, 5))

ax.scatter(jetclass_top_linear["epoch"], jetclass_top_linear["train_loss"], label="train loss linear",
           color = "red",
           marker = "^")
ax.scatter(jetclass_top_linear["epoch"], jetclass_top_linear["val_loss"], label=f"val loss linear \n min = {top_linear_min}",
           color = "red",
           marker = "s")

ax.scatter(jetclass_top_factorized["epoch"], jetclass_top_factorized["train_loss"], label="train loss factorized",
           color = "blue",
           marker = "^")
ax.scatter(jetclass_top_factorized["epoch"], jetclass_top_factorized["val_loss"], label=f"val loss factorized \n min = {top_factorized_min}",
           color = "blue",
           marker = "s")
ax.plot(jetclass_top_linear["epoch"], jetclass_top_linear["train_loss"],
           color = "red",
           linewidth = 0.5,
           alpha = 0.5)
ax.plot(jetclass_top_linear["epoch"], jetclass_top_linear["val_loss"],
           color = "red",
           linewidth = 0.5,
           alpha = 0.5)
ax.plot(jetclass_top_factorized["epoch"], jetclass_top_factorized["train_loss"],
           color = "blue",
           linewidth = 0.5,
           alpha = 0.5)
ax.plot(jetclass_top_factorized["epoch"], jetclass_top_factorized["val_loss"],
           color = "blue",
           linewidth = 0.5,
           alpha = 0.5)




ax.vlines(top_linear_min, 0, 100, color = "red", linestyles="--")
ax.vlines(top_factorized_min, 0, 100, color = "blue", linestyles="--")

ax.legend()

ax.xaxis.set_major_locator(MaxNLocator(integer=True))

#ax.set_yscale("log")
ax.set_title("Training/Validation Loss Top")

ax.set_xlim(0.2, 20)
ax.set_ylim(5.2, 5.7)
ax.set_xlabel("Epoch")
ax.set_ylabel("Loss")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

ax.scatter(jetclass_qcd_linear["epoch"], jetclass_qcd_linear["train_loss"], label="train loss linear",
           color = "red",
           marker = "^")
ax.scatter(jetclass_qcd_linear["epoch"], jetclass_qcd_linear["val_loss"], label=f"val loss linear \n min = {qcd_linear_min}",
           color = "red",
           marker = "s")

ax.scatter(jetclass_qcd_factorized["epoch"], jetclass_qcd_factorized["train_loss"], label="train loss factorized",
           color = "blue",
           marker = "^")
ax.scatter(jetclass_qcd_factorized["epoch"], jetclass_qcd_factorized["val_loss"], label=f"val loss factorized \n min = {qcd_factorized_min}",
           color = "blue",
           marker = "s")
ax.plot(jetclass_qcd_linear["epoch"], jetclass_qcd_linear["train_loss"],
           color = "red",
           linewidth = 0.5,
           alpha = 0.5)
ax.plot(jetclass_qcd_linear["epoch"], jetclass_qcd_linear["val_loss"],
           color = "red",
           linewidth = 0.5,
           alpha = 0.5)
ax.plot(jetclass_qcd_factorized["epoch"], jetclass_qcd_factorized["train_loss"],
           color = "blue",
           linewidth = 0.5,
           alpha = 0.5)
ax.plot(jetclass_qcd_factorized["epoch"], jetclass_qcd_factorized["val_loss"],
           color = "blue",
           linewidth = 0.5,
           alpha = 0.5)

ax.vlines(qcd_linear_min, 0, 100, color = "red", linestyles="--")
ax.vlines(qcd_factorized_min, 0, 100, color = "blue", linestyles="--")

ax.legend()

#ax.set_yscale("log")
ax.set_title("Training/Validation Loss QCD")

ax.xaxis.set_major_locator(MaxNLocator(integer=True))

ax.set_xlim(0.2, 20)
ax.set_ylim(4.8, 5.3)
ax.set_xlabel("Epoch")
ax.set_ylabel("Loss")
plt.show()
```
