```python
from dataset import JetDataSet
from torch.utils.data import DataLoader
from train import build_dataloader
from tqdm import tqdm
```

```python
data_loader = DataLoader(JetDataSet(
    data_dir = "/hpcwork/rwth0934/hep_foundation_model/preprocessed_data/TTBar_train_processed.h5",
    tag = "train",
    num_features=3,
    num_bins=(40, 30, 30),
    num_const=10,
    add_stop=True,
    add_start=True,
    n_jets=100,
    key = "discretized",
    ),
    batch_size = 10,
    shuffle= True, # optimization
    num_workers=4,            
    pin_memory=True,
    persistent_workers=True,
    prefetch_factor=4,)
```

```python
for epoch in range(10):
    progress_bar = tqdm(
            data_loader,
            desc = f"Epoch {epoch+1}/{10} [Training]",
            leave = True
    )
    for x in progress_bar:
        print(x)

        

```
